```python
import numpy as np

```

```python
# CASE STUDY 1 -Employees performance & salary analysis

employees = np.array(["Arun", "Bhavana", "Charan", "Divya", "Eshan", "Fathima"])

performance = np.array([
    [78, 82, 85, 88],
    [65, 70, 68, 74],
    [92, 89, 95, 91],
    [55, 62, 60, 67],
    [81, 79, 84, 87],
    [72, 75, 78, 80]
])

quarters = np.array(["Q1", "Q2", "Q3", "Q4"])
```

```python
#calculate the total performance score for every employees

total = np.sum(performance, axis=1)
print(f"{employees[0]} {total[0]}\n{employees[1]} {total[1]}\n{employees[2]} {total[2]}\n{employees[3]} {total[3]}\n{employees[4]} {total[4]}\n{employees[5]} {total[5]}")
```

```python
#calculate the average quarterly performance for every employee

avg = np.mean(performance, axis=1)
print(f"Average quarterly performance {avg}")
```

```python
#calculate the average performance for each quarter

quarter_avg = np.mean(performance, axis=0)
print(f"avg qrtrly {quarter_avg}")
```

```python
#identify the employee with the highest total performance score

total = np.sum(performance, axis=1)
index = np.argmax(total)
print(f"Highest performance {employees[index]} = {total[index]}")
```

```python
#rank all employees based on their total performance score

rank = np.argsort(total)[::-1]
print(employees[rank])
print(total[rank])
```

```python
#calculate the quarter-over-quarter percentage growth for every employeee from Q1 to Q4
```

```python
#Employees whose average performance is ≥ 80

avg = np.mean(performance, axis=1)
print(employees[avg >= 80])
```

```python
#count how many employees have an average performance below 70

avg = np.mean(performance, axis=1)
count = np.sum(avg < 70)
print(f"Employees below 70 = {count}")
```

```python
#find the highest performance score recorded in the entire dataset

highest = np.max(performance)
print(f"Highest performance score = {highest}")
```

```python
#identify the employee and quarter in which the highest performance score ocuured

highest = np.max(performance)
row, col = np.where(performance == highest)
print(f"Employee = {employees[row[0]]}")
print(f"Quarter = Q{col[0]+1}")
print(f"Score = {highest}")

```

```python
# CASE STUDY 2- Retail Store Performance Dashboard Operation


stores = np.array(["Store A", "Store B", "Store C", "Store D", "Store E"])

sales = np.array([
    [120000, 135000, 142000, 155000],
    [98000, 105000, 110000, 108000],
    [150000, 148000, 160000, 172000],
    [87000, 92000, 99000, 115000],
    [130000, 138000, 141000, 149000]
])
```

```python
# Calculate total sales for every store

total_sales = np.sum(sales, axis=1)
print(f"Total sales = {total_sales}")
```

```python
#calculate average quarterly sales for every store

average_sales = np.mean(sales, axis=1)
print(f"Average quarterly sales = {average_sales}")
```

```python
#calculate total sales for every quarter

quarter_total = np.sum(sales, axis=0)
print(f"Total sales for every quarter = {quarter_total}")
```

```python
#Identify the store with the highest annual sales

highest_store = np.argmax(total_sales)
print(f"Store with highest annual sales = {stores[highest_store]}")
```

```python
#Rank stores by annual sales

rank = np.argsort(total_sales)[::-1]
print(f"Stores ranked by annual sales = {stores[rank]}")
```

```python
#Calculate quarter-over-quarter (QoQ) percentage change
```

```python
#Flag stores whose annual sales exceed the overall average

overall_avg = np.mean(total_sales)
flag = np.where(total_sales > overall_avg, "Above Average", "Below Average")
print(f"Stores = {stores}, Status = {flag}")
```

```python
# CASE STUDY 3 -Student Performance Analytics


students = np.array(["Anu", "Binu", "Cathy", "Deepa", "Ebin", "Fahad"])

marks = np.array([
    [82, 75, 91, 88],
    [61, 68, 70, 65],
    [45, 52, 48, 55],
    [90, 94, 89, 92],
    [72, 80, 76, 85],
    [38, 42, 49, 44]
])

subjects = np.array(["Python", "SQL", "Excel", "Statistics"])
```

```python
#Calculate total and average marks for each student

total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
print(f"Total = {total}, Average = {average}")
```

```python
#Calculate subject averages

subject_avg = np.mean(marks, axis=0)
print(f"Subject averages = {subject_avg}")
```

```python
#Find the highest mark in each subject

highest = np.max(marks, axis=0)
print(f"Highest marks: Python = {highest[0]}, SQL = {highest[1]}, Excel = {highest[2]}, Statistics = {highest[3]}")
```

```python
#Create Pass/Fail status using 50 as the threshold

status = np.where(marks >= 50, "Pass", "Fail")
print(f"Pass/Fail status = {status}")
```

```python
#Identify students who passed every subject

passed_all = np.all(marks >= 50, axis=1)
print(f"Students who passed every subject = {students[passed_all]}")
```

```python
#Identify students whose average is below 60

below_60 = average < 60
print(f"Students with average below 60 = {students[below_60]}")
```

```python
#Assign grades

grade = np.where(average >= 80, "A", np.where(average >= 60, "B", np.where(average >= 50, "C", "F")))
print(f"Grades = {grade}")
```
